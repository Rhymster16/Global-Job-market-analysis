```python
def classify_experience(exp):

    exp = str(exp)

    if '0' in exp or 'fresher' in exp.lower():
        return 'Fresher'

    elif '1' in exp or '2' in exp:
        return 'Junior'

    elif '3' in exp or '4' in exp or '5' in exp:
        return 'Mid'

    else:
        return 'Senior'
```

```python
master_jobs['experience_level'] = (
    master_jobs['experience']
    .apply(classify_experience)
)
```

```python
master_jobs['experience_level'].value_counts()
```

```python
master_jobs['skill_count'] = (
    master_jobs['skills']
    .astype(str)
    .str.split(',')
    .str.len()
)
```

```python
master_jobs['skill_count'].describe()
```

```python
ai_keywords = [
    'ai',
    'artificial intelligence',
    'chatgpt',
    'openai',
    'llm',
    'langchain',
    'genai',
    'prompt engineering',
    'machine learning',
    'deep learning'
]
```

```python
def ai_score(text):

    text = str(text).lower()

    score = 0

    for word in ai_keywords:

        if word in text:
            score += 1

    return score
```

```python
master_jobs['ai_adoption_score'] = (
    master_jobs['description']
    .apply(ai_score)
)
```

```python
master_jobs['ai_flag'] = (
    master_jobs['ai_adoption_score']
    > 0
).astype(int)
```

```python
master_jobs['ai_flag'].value_counts()
```

```python
tier1 = [
    'Bengaluru',
    'Mumbai',
    'Delhi',
    'Pune',
    'Hyderabad',
    'Chennai',
    'Noida',
    'Gurugram'
]
```

```python
def city_tier(city):

    city = str(city)

    if city in tier1:
        return 'Tier 1'

    elif city == 'Unknown':
        return 'Unknown'

    else:
        return 'Tier 2/3'
```

```python
master_jobs['city_tier'] = (
    master_jobs['location']
    .apply(city_tier)
)
```

```python
master_jobs['posted_date'] = pd.to_datetime(
    master_jobs['posted_date'],
    errors='coerce'
)
```

```python
master_jobs['posting_month'] = (
    master_jobs['posted_date']
    .dt.month_name()
)
```

```python
master_jobs['posting_year'] = (
    master_jobs['posted_date']
    .dt.year
)
```

```python
import re

def extract_salary_min(text):

    try:
        nums = re.findall(r'\d+', str(text))

        if len(nums) > 0:
            return float(nums[0])

        return None

    except:
        return None
```

```python
def extract_salary_max(text):

    try:
        nums = re.findall(r'\d+', str(text))

        if len(nums) > 1:
            return float(nums[1])

        elif len(nums) == 1:
            return float(nums[0])

        return None

    except:
        return None
```

```python
master_jobs['salary_min'] = (
    master_jobs['salary']
    .apply(extract_salary_min)
)

master_jobs['salary_max'] = (
    master_jobs['salary']
    .apply(extract_salary_max)
)
```

```python
master_jobs['salary_midpoint'] = (
    master_jobs['salary_min']
    + master_jobs['salary_max']
) / 2
```

```python
master_jobs['hiring_complexity_score'] = (
    master_jobs['skill_count']
    +
    master_jobs['ai_adoption_score']
)
```
